Dictionaries - Unordered Collections of items . They store data in key-value pairs. Keys must be unique and immutable(eg. Strings,numbers, or tuples), where variables can be of any type.


```python
## Creating dictionaries

empty_dict={}
print(type(empty_dict))
```

```python
empty_dict=dict()
empty_dict
```

```python
student = {"name":"Shubhankar","age":32,"grade":24}

print(student)
type(student)
```

```python
## Error

student={"name":"Shubhankar","age":32,"name":24}

print(student)
```

```python
## Accessing dict elements

student = {"name":"Shubhankar","age":32,"grade":'A'}

print(student["grade"])

## Accessing using get() method
print(student.get('age'))
print(student.get("last_name","not available"))
```

Modifying dictionary Elements

```python
print(student)
```

```python
student["age"]=22 #Updated the value
print(student)
student["address"]="India" #Added new key & value
print(student)
```

```python
del student['grade'] #Delete key value pair

print(student)
```

Dict Methods

```python
keys = student.keys() #get all the keys
print(keys)
values=student.values() #get all the values
print(values)

items=student.items() #Get all key-value pairs
print(items)
```

```python
# Shallow copy

student_copy = student
print(student_copy)
```

```python
student["name"]="Shubh"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy()
print(student_copy1)
print(student)
```

```python
student["name"]="Shubh2"
```

```python
print(student_copy1)
print(student)
```

Iterating over dicts

```python
#Iterating over keys

for keys in student.keys():
    print(keys)
```

```python
#Iterating over values

for values in student.values():
    print(values)
```

```python
#Iterating over key-value pairs

for key,value in student.items():
    print(f"{key}:{value}")
```

### Nested dict

```python
students={
    "student1":{"name":"krish","age":22},
    "student2":{"name":"Peter","age":32},    
}
print(students)
```

```python
# Access nested dict elements

print(students["student2"]["name"])
print(students["student2"]["age"])
```

```python
# Iterating over nested dict 

for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dict comprehension

squares={ x:x**2 for x in range(5)}
print(squares)
```

```python
# Conditional dict comprehension

evens={x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
##  Practical Examples

## Use a dict to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]

frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number] = 1
print(frequency)
```

```python
## Merge 2 dict into one

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}

merged_dict={**dict1,**dict2}
print(merged_dict)
```
