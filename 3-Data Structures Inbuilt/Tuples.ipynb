```python
## Packing & Unpacking

# Packing

packed_tuple = 1, "Hello", 3.14
print(packed_tuple)
print("-----")

#Unpacking
a,b,c = packed_tuple
print(a)
print(b)
print(c)
print("-----")
```

```python
# Unpacking with *
numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
nested_tuple = ((1,2,3),("a","b","c"),(True,False))

print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
