Function is a block of code that performs a specific task. It helps in organizing code, reusing code, and improve readability.

```python
# syntax

def func_name(parameters):
    """Docstring"""
    # Func body
    return expression
```

```python
## Why func

num = 24

if num%2==0:
    print("the no is even")
else:
    print("the no is odd")
```

```python
def even_or_odd(num):
    """This func finds even or odd"""
    if num%2==0:
        print(f"{num} - is even.")
    else:
        print(f"{num} - is odd.")
```

```python
## Call this function

even_or_odd(25)
```

```python
## Function with multiple parameters

def add(a,b):
    """This func adds two numbers"""
    return a+b

result = add(2,4)
print(result)
```

```python
## Default Parameters

def greet(name="Guest"):
    print(f"Hello {name}")

greet()
greet("Shubhankar")
```

```python
## Variable Length Arguments

## Positional arguments

def print_numbers(*args): ## *args is positional arguments
    for numbers in args:
        print(numbers)
```

```python
print_numbers(1,2,3,4,5,6,"Shubh")
```

```python
## Keywords arguments :: All parameters will be in the form of key-value pairs

def print_details(*args,**kwargs): ## **kwargs is keyword arguments

    for val in args:
        print(f"Positional Argument:{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,"Shubh",name="Shubhankar",age="22",country="India")
```

```python
## Return multiple parameters

def multiply(a,b):

    return a*b,a

multiply(2,3)
```
