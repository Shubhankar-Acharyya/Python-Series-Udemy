## Exceptions:
Events that disrupt the normal flow of a program. They occur when an error is encountered during program execution. 
Common exceptions::
1. ZeroDivisionError
2. FileNotFoundError
3. ValueError
4. TypeError

```python
## Exeption try, except block

try:
    a=b
except:
    print("The Variable has not been assigned.")
```

```python
try:
    result=1/0
    a=b
except ZeroDivisionError as ex:
    print(ex)
except Exception as ex1:
    print(ex1)
```

```python
try:
    num=int(input("Enter a number: "))
    result=10/num
except ValueError:
    print("Please enter a valid number.")
except ZeroDivisionError:
    print("Enter denominator greater than zero.")
except Exception as ex:
    print(ex)
```

```python
## try,except,else block
try:
    num=int(input("Enter a number: "))
    result=10/num
except ValueError:
    print("Please enter a valid number.")
except ZeroDivisionError:
    print("Enter denominator greater than zero.")
except Exception as ex:
    print(ex)
else:
    print("The result is: ",result)
```

```python
### try,except,else,finally block


try:
    num=int(input("Enter a number: "))
    result=10/num
except ValueError:
    print("Please enter a valid number.")
except ZeroDivisionError:
    print("Enter denominator greater than zero.")
except Exception as ex:
    print(ex)
else:
    print("The result is: ",result)
finally:
    print("Execution completed.")
```

```python
## File handling and Exception handling

try:
    file=open("test.txt","r")
    data=file.read()
    print(data)
except FileNotFoundError:
    print("File not found.")
except Exception as ex:
    print(ex)
finally:
    try:
        file.close()
    except:
        pass
```
