```python
### read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
## Read a file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line.strip())  ##.strip() removes new line
```

```python
## Writing a file(Overwriting)

with open('example.txt','w') as file:
    file.write('Hello World!\n')
    file.write('this is a new line.')
```

```python
## Write a file(without Overwriting)
with open('example.txt','a') as file:
    file.write("\nAppend Operation taking place")
```

```python
### Writing a list of lines to a file

lines =['First line\n','Second line \n','Third line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
### Writing a Binary Files

data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
### Reading a Binary Files


with open('example.bin','rb') as file:
    content = file.read()
    print(content)
```

```python
### Read the content from source text file and write to a dest text file

## Copying a text file
with open('example.txt','r') as source_file:
    content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
```

```python
# Read a txt file and count the no of lines, words and chars.

## Counting lines, words and chars in a text file
def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path ='example.txt'
lines,words,chars = count_text_file(file_path)
print(f'lines:{lines},words:{words},Characters:{chars}')
```

```python
## Writing and then reading a file

with open('example.txt','w+') as file:
    file.write("Hello World\n")
    file.write("This is a new Line\n")

    ## Move the file cursor to the beginning
    file.seek(0)

    # Read the content of the file
    content=file.read()
    print(content)
```

```python
path='example.txt'
if os.path.exists(path):
    print(f'The file {path} exists.')
else:
    print(f'The file {path} does not exist.')
```

```python
# Check if a path is a file or directory

import os

path = 'example.txt'
if os.path.isfile(path):    
    print(f'The path {path} is a file.')
elif os.path.isdir(path):
    print(f'The path {path} is a directory.')
else:
    print(f'The path {path} is neither a file nor a directory.')
```

```python
### Getting the absolute path of a file

relative_path = 'example.txt'
absolute_path = os.path.abspath(relative_path)
print(f'The absolute path of {relative_path} is: {absolute_path}')
```
