#### Importing
In python importing help organize and reuse code.

```python
import math
math.sqrt(16)
```

```python
from math import sqrt,pi

print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
from math import *
print(sqrt(16))
print(pi)
```
