for loop

```python
for i in range(1,6):
    print(i)
```

```python
for i in range(10,1,-2): #-2 is step value
    print(i)
```

while loop - continues to execute as long as condition is true

```python
count = 0

while count < 5:
    print(count)
    count += 1
```

```python
count = 0

while count%2==0:
    print(count)
    count += 1
```

## Loop Control Statements

break - exits the loop prematurely

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

continue - skips current and continues to next

```python
for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

pass - null operation and does nothing

```python
for i in range(5):
    if i == 3:
        pass
    print(i)
```

## Nested Loops

```python
for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
## Examples - Calculate the sum of first 10 natural numbers using for loop and while loop

# Using while loop

n = 10
sum = 0
count = 1

while count <= n:
    sum += count
    count += 1

print(f"The sum of first {n} natural numbers is: {sum}")
```

```python
# Using for loop

n = 10
sum = 0
count = 1

for i in range(1, n + 1):
    sum += i

print(f"The sum of first {n} natural numbers is: {sum}")
```

```python
## Example - Prime numbers bw 1 and 100

for num in range(1, 101):
    if num > 1:
        for i in range(2, num):
            if (num % i) == 0:
                break
        else:
            print(num)
```
