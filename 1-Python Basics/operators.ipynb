Arithmetic Operators

```python
a = 10
b = 5

add_result=a+b
sub_result=a-b
mul_result=a*b
div_result=a/b
floor_div_result=a//b
modulus_result=a%b
exponent_result=a**b

print(add_result)
print(sub_result)
print(mul_result)
print(div_result)
print(floor_div_result)
print(modulus_result)
print(exponent_result)
```

```python
# ==
a=10
b=10

a==b
```

```python
# >

num1 = 10
num2 = 15

num1>num2
```

```python
# <

num1<num2
```

```python
# >=

num1 = 45
num2 = 45

num1>=num2
```

Logical Operators

AND Operator

```python

X = True
Y = True

result = X and Y
print(result)
```

```python
X = True
Y = False

result = X and Y
print(result)
```

OR Operator

```python
X = True
Y = True

result = X or Y
print(result)
```

```python
X = False
Y = False

result = X or Y
print(result)
```

```python
X = True

not X
```
