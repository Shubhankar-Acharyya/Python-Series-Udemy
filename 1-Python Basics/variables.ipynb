Basic Calculator

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

sum = num1 + num2
sub = num1 - num2
mul = num1 * num2
div = num1 / num2

print("Sum:", sum)
print("Sub:", sub)
print("Mul:", mul)
print("Div:", div)
```
